```python
data.purpose.value_counts()
purpose_map = {
    "debt_consolidation": "Debt Management",
    "credit_card": "Debt Management",
    
    "medical": "Personal Expenses",
    "vacation": "Personal Expenses",
    "moving": "Personal Expenses",
    "wedding": "Personal Expenses",
    "other": "Personal Expenses",
    
    "home_improvement": "Home & Property",
    "house": "Home & Property",
    "renewable_energy": "Home & Property",
    
    "car": "Vehicle & Major Purchases",
    "major_purchase": "Vehicle & Major Purchases",
    
    "small_business": "Business & Education"
}


data["purpose_group"] = data["purpose"].map(purpose_map)
data = data.drop('purpose',axis=1)
data.purpose_group.value_counts()
```

```python
data['delinq_2yrs_flag']= data.delinq_2yrs.clip(0,1)
data = data.drop('delinq_2yrs',axis =1)
```

```python
data.inq_last_6mths
bins = [-np.inf, 0, 1, 2, np.inf]   
labels = ["0", "0-1", "1-2", "2+"]   

data["inq_last_6mths_bin"] = pd.cut(data["inq_last_6mths"], 
                                  bins=bins, 
                                  labels=labels)
data = data.drop('inq_last_6mths',axis =1 )
```

```python
import numpy as np
import pandas as pd

# total_acc_bin
bins = [-np.inf, 10, 16, 20, 26, 35, np.inf]
labels = ["0-10", "10-16", "16-20", "20-26", "26-35", "35+"]
data["total_acc_bin"] = pd.cut(data["total_acc"], bins=bins, labels=labels)
data = data.drop("total_acc", axis=1)

# bin_annual_inc
bins = [-np.inf, 18000, 40000, 80000, 120000, np.inf]
labels = ["<18K", "18K-40K", "40K-80K", "80K-120K", ">120K"]
data["bin_annual_inc"] = pd.cut(data["annual_inc"], bins=bins, labels=labels)
data = data.drop("annual_inc", axis=1)

# bin_int_ratec
bins = [-np.inf, 8, 12, 15, 21, 24, np.inf]
labels = ["<8%", "8-12%", "12-15%", "15-21%", "21-24%", ">24%"]
data["bin_int_ratec"] = pd.cut(data["int_rate"], bins=bins, labels=labels)
data = data.drop("int_rate", axis=1)
```

```python
data["emp_status_unverifiable"] = np.where(data.emp_length =='UNKNOWN',0,1)
data = data.drop('emp_length',axis =1)
```

```python

# bin_dti
bins = [-np.inf, 5, 10, 15, 20, 25, 30, 35, np.inf]
labels = ["<5", "5-10", "10-15", "15-20", "20-25", "25-30", "30-35", ">35"]

data["bin_dti"] = pd.cut(data["dti"], 
                         bins=bins, 
                         labels=labels)
data = data.drop("dti", axis=1)

# bin_revol_util
bins = [-np.inf, 20, 40, 60, 80, 100, np.inf]
labels = ["<20%", "20-40%", "40-60%", "60-80%", "80-100%", ">100%"]

data["bin_revol_util"] = pd.cut(data["revol_util"], 
                                bins=bins, 
                                labels=labels)
data = data.drop("revol_util", axis=1)
```

```python
data = data[['loan_amnt', 'term', 'sub_grade', 'home_ownership',
       'verification_status', 'purpose_group', 'loan_status',
       'delinq_2yrs_flag', 'inq_last_6mths_bin', 'total_acc_bin',
       'emp_status_unverifiable', 'bin_annual_inc', 'bin_int_ratec', 'bin_dti',
       'bin_revol_util','loan_status']]
```
